```python
def latlng_to_local_m(lat, lng, ref_lat, ref_lng):
    m_per_deg_lat = 111320.0
    m_per_deg_lng = 111320.0 * np.cos(np.radians(ref_lat))
    east = (lng - ref_lng) * m_per_deg_lng
    north = (lat - ref_lat) * m_per_deg_lat
    return east, north

def local_m_to_latlng(east, north, ref_lat, ref_lng):
    m_per_deg_lat = 111320.0
    m_per_deg_lng = 111320.0 * np.cos(np.radians(ref_lat))
    lat = ref_lat + north / m_per_deg_lat
    lng = ref_lng + east / m_per_deg_lng
    return lat, lng
```

```python
def load_telemetry(path):
    df = pd.read_csv(path)
    df = df[["time_s", "lat", "lng", "height_m", "yaw_deg"]].dropna().reset_index(drop=True)
    return df

def get_drone_state(telemetry, time_s):
    idx = (telemetry["time_s"] - time_s).abs().idxmin()
    row = telemetry.loc[idx]
    return row["lat"], row["lng"], row["height_m"], row["yaw_deg"]
```

## Ground Control Points (GCPs)

The corner names are specified by azimuth from the center of the quadrilateral (they do not depend on the drone's heading):
`corner_A_ESE`, `corner_B_SE`, `corner_C_WNW`, `corner_D_NW`.


```python
GCP_FRAME = 741

GROUND_CONTROL_POINTS = [
    {"corner": "corner_D_NW",  "px": 500,  "py": 220, "lat": 52.75236395813839, "lon": 13.680075477949087},
    {"corner": "corner_A_ESE", "px": 1650, "py": 600, "lat": 52.75142022264314, "lon": 13.68265678136193},
    {"corner": "corner_B_SE",  "px": 1050, "py": 630, "lat": 52.75073848839966, "lon": 13.6822158755006},
    {"corner": "corner_C_WNW", "px": 0,    "py": 250, "lat": 52.75163129122488, "lon": 13.6794141191636},
]

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

ax = axes[0]
pxs = [g["px"] for g in GROUND_CONTROL_POINTS] + [GROUND_CONTROL_POINTS[0]["px"]]
pys = [g["py"] for g in GROUND_CONTROL_POINTS] + [GROUND_CONTROL_POINTS[0]["py"]]
ax.plot(pxs, pys, "o-", color="steelblue")
for i, g in enumerate(GROUND_CONTROL_POINTS):
    ax.annotate(f"{i+1}. {g['corner']}", (g["px"], g["py"]), fontsize=9, xytext=(8, 8), textcoords="offset points")
ax.set_xlim(-100, IMG_WIDTH + 100); ax.set_ylim(IMG_HEIGHT + 100, -100)
ax.set_xlabel("px"); ax.set_ylabel("py"); ax.grid(alpha=0.3)

ax = axes[1]
lons = [g["lon"] for g in GROUND_CONTROL_POINTS] + [GROUND_CONTROL_POINTS[0]["lon"]]
lats = [g["lat"] for g in GROUND_CONTROL_POINTS] + [GROUND_CONTROL_POINTS[0]["lat"]]
ax.plot(lons, lats, "o-", color="darkorange")
for i, g in enumerate(GROUND_CONTROL_POINTS):
    ax.annotate(f"{i+1}. {g['corner']}", (g["lon"], g["lat"]), fontsize=9, xytext=(8, 8), textcoords="offset points")
ax.set_title("GPS")
ax.set_xlabel("lon"); ax.set_ylabel("lat"); ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig("gcp_check.png", dpi=110)
plt.show()
```

```python
# Independent estimation of a drone's pose using 4 GCPs (without telemetry)
def project_ray_to_ground(drone_east, drone_north, height, yaw_deg, pitch_down_deg, focal_px, px, py):
    yaw = np.radians(yaw_deg)
    pitch = np.radians(pitch_down_deg)
    forward = np.array([np.sin(yaw) * np.cos(pitch), np.cos(yaw) * np.cos(pitch), -np.sin(pitch)])
    right   = np.array([np.cos(yaw), -np.sin(yaw), 0])
    down    = np.cross(forward, right)
    dx_px, dy_px = px - IMG_WIDTH / 2, py - IMG_HEIGHT / 2
    direction = right * (dx_px / focal_px) + down * (dy_px / focal_px) + forward
    direction = direction / np.linalg.norm(direction)
    if direction[2] >= -1e-6:
        return None
    t = -height / direction[2]
    return drone_east + t * direction[0], drone_north + t * direction[1]

def fit_single_frame_pose(gcps, ref_lat, ref_lng):
    def residuals(params):
        drone_east, drone_north, height, yaw_deg, pitch_down_deg, focal_px = params
        errs = []
        for g in gcps:
            proj = project_ray_to_ground(drone_east, drone_north, height, yaw_deg, pitch_down_deg, focal_px, g["px"], g["py"])
            if proj is None:
                errs.extend([1000, 1000]); continue
            pe, pn = proj
            te, tn = latlng_to_local_m(g["lat"], g["lon"], ref_lat, ref_lng)
            errs.extend([pe - te, pn - tn])
        return errs

    best, best_rmse = None, np.inf
    for pitch_guess in [90, 80, 70, 60, 50, 40, 30]:
        x0 = [0, 0, 25.0, 0.0, pitch_guess, 1200.0]
        res = least_squares(residuals, x0=x0, max_nfev=20000,
                             bounds=([-500, -500, 5, -180, 5, 200], [500, 500, 150, 180, 90, 5000]))
        errs = np.array(residuals(res.x)).reshape(-1, 2)
        rmse = np.sqrt((errs**2).sum(axis=1)).mean()
        if rmse < best_rmse:
            best_rmse, best = rmse, res
    return best, best_rmse

REF_LAT = np.mean([g["lat"] for g in GROUND_CONTROL_POINTS])
REF_LNG = np.mean([g["lon"] for g in GROUND_CONTROL_POINTS])

pose_fit, pose_rmse = fit_single_frame_pose(GROUND_CONTROL_POINTS, REF_LAT, REF_LNG)
drone_east, drone_north, height_est, yaw_est, pitch_est, focal_est = pose_fit.x

print(f"RMSE calibration: {pose_rmse:.2f} m")
print(f"Height: {height_est:.1f} m | Course: {yaw_est:.1f}° | Camera pitch: {pitch_est:.1f}° | Focus: {focal_est:.0f}px")
```

```python
# Leave-one-out
for i in range(len(GROUND_CONTROL_POINTS)):
    subset = [g for j, g in enumerate(GROUND_CONTROL_POINTS) if j != i]
    ref_lat = np.mean([g["lat"] for g in subset])
    ref_lng = np.mean([g["lon"] for g in subset])
    b, r = fit_single_frame_pose(subset, ref_lat, ref_lng)
    print(f" without {GROUND_CONTROL_POINTS[i]['corner']:14s} -> RMSE={r:6.2f}m  pitch={b.x[4]:5.1f}°  height={b.x[2]:5.1f}m")
```

```python
video_time_gcp = GCP_FRAME / VIDEO_FPS

results = []
for fname in TELEMETRY_FILES:
    telemetry = load_telemetry(fname)
    file_duration = telemetry["time_s"].max()
    max_offset = max(0.0, file_duration - VIDEO_DURATION_S)
    offsets = np.arange(0, max_offset + 0.5, 0.5)

    best = None
    for off in offsets:
        t_query = video_time_gcp + off
        lat, lng, h, yaw = get_drone_state(telemetry, t_query)
        e_t, n_t = latlng_to_local_m(lat, lng, REF_LAT, REF_LNG)
        pos_err = np.hypot(e_t - drone_east, n_t - drone_north)
        h_err = abs(h - height_est)
        yaw_err = abs(((yaw - yaw_est + 180) % 360) - 180)
        combined = pos_err + h_err * 2 + yaw_err * 0.3
        if best is None or combined < best[0]:
            best = (combined, off, pos_err, h_err, yaw_err)

    results.append((fname, *best))

results.sort(key=lambda x: x[1])
print(f"{'file':45s} {'offset,s':>9} {'pos_err,m':>10} {'h_err,m':>9} {'yaw_err,°':>10}")
for fname, combined, off, pos_err, h_err, yaw_err in results:
    print(f"{fname:45s} {off:9.1f} {pos_err:10.1f} {h_err:9.1f} {yaw_err:10.1f}")

best_file, best_combined, best_offset = results[0][0], results[0][1], results[0][2]
print(f"\n Best: {best_file}, offset={best_offset:.1f}s")
```

```python
# Bundle adjustment
TELEMETRY_CSV = best_file
VIDEO_OFFSET_S = best_offset

FIELD_CORNERS = {g["corner"]: {"lat": g["lat"], "lon": g["lon"]} for g in GROUND_CONTROL_POINTS}

OBSERVATIONS = [
    {"frame": 741,   "corner": "corner_D_NW",  "px": 500,  "py": 220},
    {"frame": 741,   "corner": "corner_B_SE",  "px": 1050, "py": 630},
    {"frame": 741,   "corner": "corner_A_ESE", "px": 1650, "py": 600},
    {"frame": 741,   "corner": "corner_C_WNW", "px": 0,    "py": 250},
    {"frame": 1100,  "corner": "corner_D_NW",  "px": 50,   "py": 200},
    {"frame": 1100,  "corner": "corner_A_ESE", "px": 1625, "py": 650},
    {"frame": 1500,  "corner": "corner_C_WNW", "px": 600,  "py": 250},
    {"frame": 1500,  "corner": "corner_D_NW",  "px": 1025, "py": 250},
    {"frame": 2100,  "corner": "corner_C_WNW", "px": 1550, "py": 250},
    {"frame": 2100,  "corner": "corner_B_SE",  "px": 100,  "py": 950},
    {"frame": 22150, "corner": "corner_B_SE",  "px": 150,  "py": 0},
    {"frame": 470,   "corner": "corner_A_ESE", "px": 1750, "py": 720},
    {"frame": 470,   "corner": "corner_B_SE",  "px": 600,  "py": 850},
    {"frame": 470,   "corner": "corner_D_NW",  "px": 425,  "py": 425},
]

telemetry_fixed = load_telemetry(TELEMETRY_CSV)

def get_state_by_frame(frame):
    t = frame / VIDEO_FPS + VIDEO_OFFSET_S
    return get_drone_state(telemetry_fixed, t)

def bundle_residuals(params, obs_list):
    pitch_down_deg, focal_px = params
    errs = []
    for obs in obs_list:
        lat, lng, h, yaw = get_state_by_frame(obs["frame"])
        de, dn = latlng_to_local_m(lat, lng, REF_LAT, REF_LNG)
        proj = project_ray_to_ground(de, dn, h, yaw, pitch_down_deg, focal_px, obs["px"], obs["py"])
        if proj is None:
            errs.extend([1000, 1000]); continue
        pe, pn = proj
        c = FIELD_CORNERS[obs["corner"]]
        te, tn = latlng_to_local_m(c["lat"], c["lon"], REF_LAT, REF_LNG)
        errs.extend([pe - te, pn - tn])
    return errs

result = least_squares(bundle_residuals, x0=[max(pitch_est, 5), focal_est], args=(OBSERVATIONS,),
                        max_nfev=20000, bounds=([5, 200], [90, 5000]))
PITCH_DOWN_DEG, FOCAL_PX = result.x
print(f"Final: pitch_down={PITCH_DOWN_DEG:.1f}°  focal_px={FOCAL_PX:.0f}px\n")
```

```python
# Final projection: frame pixel on GPS
def pixel_to_latlng(px, py, frame):
    lat, lng, h, yaw = get_state_by_frame(frame)
    de, dn = latlng_to_local_m(lat, lng, REF_LAT, REF_LNG)
    proj = project_ray_to_ground(de, dn, h, yaw, PITCH_DOWN_DEG, FOCAL_PX, px, py)
    if proj is None:
        return None
    return local_m_to_latlng(proj[0], proj[1], REF_LAT, REF_LNG)

# A quick check at one of the GCP points
test = GROUND_CONTROL_POINTS[0]
result_latlng = pixel_to_latlng(test["px"], test["py"], GCP_FRAME)
print(f"Checking for {test["corner"]}: received {result_latlng}, expected ({test["lat"]:.6f}, {test["lon"]:.6f})")
```

```python
# Apply this to all tracks in tracks_dense.xml
TRACKS_XML = "/kaggle/input/datasets/natair/chicken-trainings-data/tracks_dense.xml"

import os
if not os.path.exists(TRACKS_XML):
    print(f"{TRACKS_XML} not found in the working directory")
else:
    tree = ET.parse(TRACKS_XML)
    root = tree.getroot()

    rows = []
    for track in root.findall(".//track"):
        track_id = track.get("id")
        for box in track.findall("box"):
            if box.get("outside") == "1":
                continue
            frame = int(box.get("frame"))
            cx = (float(box.get("xtl")) + float(box.get("xbr"))) / 2
            cy = (float(box.get("ytl")) + float(box.get("ybr"))) / 2
            latlng = pixel_to_latlng(cx, cy, frame)
            if latlng is None:
                continue
            lat, lon = latlng
            rows.append({"track_id": track_id, "frame": frame, "time_s": frame / VIDEO_FPS, "lat": lat, "lon": lon})

    df_geo = pd.DataFrame(rows)
    df_geo.to_csv("chicken_gps_tracks.csv", index=False)
    print(f" Geolocated detections: {len(df_geo)}  |  Unique tracks: {df_geo["track_id"].nunique()}")
```

```python
# Validation: field map + chicken tracks + speed check
import os
fig, ax = plt.subplots(figsize=(8, 8))

corner_e, corner_n = [], []
for g in GROUND_CONTROL_POINTS:
    e, n = latlng_to_local_m(g["lat"], g["lon"], REF_LAT, REF_LNG)
    corner_e.append(e); corner_n.append(n)
corner_e.append(corner_e[0]); corner_n.append(corner_n[0])
ax.plot(corner_e, corner_n, "o-", color="black", label="field edges (GCP)")

if os.path.exists("chicken_gps_tracks.csv"):
    df_geo = pd.read_csv("chicken_gps_tracks.csv")
    for tid, grp in df_geo.groupby("track_id"):
        e, n = latlng_to_local_m(grp["lat"].values, grp["lon"].values, REF_LAT, REF_LNG)
        ax.plot(e, n, "-", alpha=0.6, linewidth=1)

    speeds = []
    for tid, grp in df_geo.sort_values("time_s").groupby("track_id"):
        e, n = latlng_to_local_m(grp["lat"].values, grp["lon"].values, REF_LAT, REF_LNG)
        dt = np.diff(grp["time_s"].values)
        dist = np.hypot(np.diff(e), np.diff(n))
        valid = dt > 0
        speeds.extend((dist[valid] / dt[valid]).tolist())
    speeds = np.array(speeds)
    if len(speeds):
        print(f"Track speeds: median={np.median(speeds):.2f} m/s, 95th percentile={np.percentile(speeds,95):.2f} m/s")
else:
    print("chicken_gps_tracks.csv not found")

ax.set_title("Field + chicken tracks")
ax.set_xlabel("East, m"); ax.set_ylabel("North, m")
ax.legend(); ax.grid(alpha=0.3); ax.set_aspect("equal")
plt.tight_layout()
plt.savefig("validation_map.png", dpi=110)
plt.show()
```
